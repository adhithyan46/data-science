```python
import nltk
nltk.download("punkt")
nltk.download("wordnet")
nltk.download('stopword')
nltk.download("punkt_tab")
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()

phrase = "i am studing "
words = word_tokenize(phrase)
stemmed = []
for word in words:
  stemmed.append(stemmer.stem(word))
" ".join(stemmed)
```

```python
from nltk.stem.wordnet import WordNetLemmatizer
from nltk.tokenize import word_tokenize
lemmatizer = WordNetLemmatizer()

phrase = " i am running late"
words = word_tokenize(phrase)

lemmatized = []
for word in words:
  lemmatized.append(lemmatizer.lemmatize(word,pos = "v"))
" ".join(lemmatized)
```

## Stop word removal

```python
nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

stop_words = stopwords.words("english")
print( stop_words)
```

```python
phrase = "I am going to watch movie with my friends on bike "
words = word_tokenize(phrase)

No_stopword = []
removed_words = []
for w in words:
  if w not in stop_words:
    No_stopword.append(w)
  else:
    removed_words.append(w)

print(f'Without stopwords ->'," ".join(No_stopword))
print(f'Removed stopwords ->'," ".join(removed_words))
```
