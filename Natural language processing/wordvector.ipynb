# Word Vector Similarity

```python
import spacy
nlp = spacy.load("en_core_web_lg")
```



*   tok.text - original text
*   tok.has_vector - has the text have vector ? ( true or false )
*   tok.norm - The square root of sum of value squared
*   tok.oov - out of vocabulary






```python
token = nlp("dog cat banana asdfas")
for tok in token :
  print(tok.text, tok.has_vector, tok.vector_norm, tok.is_oov)
```

```python
doc1 = nlp("I love nakedsport motorbikes ")
doc2 = nlp("Adventure motorcycles are also good ")

#similarity of 2 documents
print(doc1, "<->" ,doc2,doc1.similarity(doc2) )
```

```python
bike = doc1[3]
cycle = doc2[1:2]
print(bike, "<->", cycle, bike.similarity(cycle)) #similarity between 2 words
```

# Making Predictions

```python
class Category :
  Adv = "adv"
  sports = "sports"

x_train = ["enduro goes through mud and offroad", "ADvs are capable of going long touring ", "sports bikes can go on tracks ","we can have fun in race track but not in street"]
y_train = [Category.Adv,Category.Adv, Category.sports, Category.sports]
```

```python
# Word vectorization
doc = [nlp(text) for text in x_train]
train_wv = [x.vector for x in doc]
# print(train_wv[0].vector)
```

```python
from sklearn import svm
clf_svm_wv = svm.SVC(kernel = "linear")
clf_svm_wv.fit(train_wv, y_train)
```

```python
x_test = (["how to tour"])
test_doc = [nlp(text) for text in x_test]
test_x_wv = [x.vector for x in test_doc]
clf_svm_wv.predict(test_x_wv)
```

# Regexes ( Regular Expression )

```python
import re
regexp = re.compile(r"^adh[^\s]*yan$")
phrases = ["adhithyan","adithyan","absdfascd","abasdfa"]
matches = []
for phrase in phrases:
  if re.match(regexp,phrase):
    matches.append(phrase)

print(matches)
```
