```python
import nltk
nltk.download("punkt")
nltk.download('stopword')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
phrase = 'The children were running quickly towards the buildings, hoping to find better shelters before the storm worsened.'
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()
stemmed_words = []
words = word_tokenize(phrase)
for word in words :
  stemmed_words.append(stemmer.stem(word))

" ".join(stemmed_words)
```

```python
from nltk.stem.wordnet import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = []

words = word_tokenize(phrase)

for word in words :
  lemmatized_words.append(lemmatizer.lemmatize(word))

" ".join(lemmatized_words)
```

```python
phrase = 'The children were running quickly towards the buildings, hoping to find better shelters before the storm worsened.'
```

```python
from nltk.stem.wordnet import WordNetLemmatizer
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
import spacy

nlp = spacy.load("en_core_web_sm")

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

words = word_tokenize(phrase)
# print("Original words:", words)

lemmatized = [lemmatizer.lemmatize(word) for word in words]
print("Lemmatized :",lemmatized)

stemmed = [stemmer.stem(word) for word in words]
# print("Stemmed :", stemmed)

doc = nlp(phrase)
lemmatized_spacy = [token.lemma_ for token in doc]
print("Lemmatized with spacy :", lemmatized_spacy)
```
