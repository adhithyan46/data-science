```python
import spacy
import torch
nlp = spacy.load("en_core_web_md")
doc = nlp("some text to encode ")
```

```python
class category :
  book = "book"
  bank = "bank"

train_x = [" i like to write ", "i love books "," check the books ", "save money", "deposite in bank", " write check for take money", "read book"]
train_y = [category.book, category.book, category.book, category.bank, category.bank, category.bank, category.book]
```

```python
from sklearn import svm

docs = [nlp(text) for text in train_x ]
train_x_vector = [doc.vector for doc in docs]
clf_svm = svm.SVC(kernel = "linear")

clf_svm.fit(train_x_vector,train_y)
```

```python
test_x  =['write check']
docs = [nlp(text) for text in test_x]
test_x_vector = [doc.vector for doc in docs]

clf_svm.predict(test_x_vector)
```
